## Scratch space

Colour distribution per class — a first look at whether the erythema signal
the Phase 5 features rely on is actually separable in this data.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

train_dir = PROCESSED / "infection" / "train"
fig, ax = plt.subplots(figsize=(9, 4))
for class_dir in sorted(p for p in train_dir.iterdir() if p.is_dir()):
    a_values = []
    for path in list(class_dir.iterdir())[:60]:
        img = cv2.imread(str(path))
        if img is None:
            continue
        lab = cv2.cvtColor(img, cv2.COLOR_BGR2LAB)
        a_values.append(lab[:, :, 1].mean())  # a* = green-red axis
    if a_values:
        ax.hist(a_values, bins=20, alpha=0.5, label=class_dir.name)
ax.set_xlabel("mean a* (LAB) — higher is redder")
ax.set_ylabel("images")
ax.legend()
plt.show()
```
